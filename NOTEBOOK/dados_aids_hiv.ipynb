```python
# Informe o ano que deseja analisar
ano_selecionado = int(input("Digite o ano que deseja analisar (ex: 2020): "))

# ============================================================
# PREPARAÇÃO DOS DADOS
# ============================================================

# Converte a coluna de data
data["dt_notific"] = pd.to_datetime(
    data["dt_notific"],
    errors="coerce"
)

# Remove registros sem data de notificação
dados = data.dropna(subset=["dt_notific"]).copy()

# Cria ano e semana ISO
iso = dados["dt_notific"].dt.isocalendar()

dados["ano_iso"] = iso.year
dados["semana_iso"] = iso.week


# ============================================================
# FILTRO DO ANO
# ============================================================

dados_ano = dados[
    dados["ano_iso"] == ano_selecionado
].copy()


# ============================================================
# CASOS POR SEMANA
# ============================================================

casos_por_semana = (
    dados_ano
    .groupby("semana_iso")
    .size()
    .reindex(range(1, 54), fill_value=0)
)

# Remove semanas que não existem no ano selecionado
casos_por_semana = casos_por_semana[
    casos_por_semana.index <= dados_ano["semana_iso"].max()
]


# ============================================================
# CASOS ACUMULADOS
# ============================================================

casos_acumulados = casos_por_semana.cumsum()


# ============================================================
# GRÁFICO 1 — CASOS POR SEMANA
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    casos_por_semana.index,
    casos_por_semana.values,
    marker="o"
)

plt.title(
    f"Casos de HIV por semana — {ano_selecionado}"
)

plt.xlabel("Semana ISO")
plt.ylabel("Número de casos")

plt.xticks(casos_por_semana.index)

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# GRÁFICO 2 — CASOS ACUMULADOS
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    casos_acumulados.index,
    casos_acumulados.values,
    marker="o"
)

plt.title(
    f"Casos acumulados de HIV por semana — {ano_selecionado}"
)

plt.xlabel("Semana ISO")
plt.ylabel("Casos acumulados")

plt.xticks(casos_acumulados.index)

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# RESUMO NO TERMINAL
# ============================================================

print(f"\nAno analisado: {ano_selecionado}")
print(f"Total de casos: {len(dados_ano)}")
print(f"Semanas analisadas: {casos_por_semana.index.min()} "
      f"até {casos_por_semana.index.max()}")

print("\nCasos por semana:")
print(casos_por_semana)

print("\nCasos acumulados:")
print(casos_acumulados)

media_movel = casos_por_semana.rolling(8).mean()

plt.figure(figsize=(14, 6))

plt.plot(
    casos_por_semana.index,
    casos_por_semana.values,
    marker="o",
    alpha=0.5,
    label="Casos por semana"
)

plt.plot(
    media_movel.index,
    media_movel.values,
    linewidth=3,
    label="Média móvel — 8 semanas"
)

plt.title(f"Casos de HIV por semana — {ano_selecionado}")

plt.xlabel("Semana ISO")
plt.ylabel("Número de casos")

plt.xticks(casos_por_semana.index)

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# CONFIGURAÇÃO
# ============================================================

# Número de semanas da média móvel
JANELA_MEDIA = 16


# ============================================================
# PREPARAÇÃO DOS DADOS
# ============================================================

# Converte a data de notificação
data["dt_notific"] = pd.to_datetime(
    data["dt_notific"],
    errors="coerce"
)

# Remove registros sem data
dados = data.dropna(subset=["dt_notific"]).copy()

# Obtém ano e semana ISO
iso = dados["dt_notific"].dt.isocalendar()

dados["ano_iso"] = iso.year
dados["semana_iso"] = iso.week


# ============================================================
# CASOS POR ANO E SEMANA
# ============================================================

casos_semanais = (
    dados
    .groupby(["ano_iso", "semana_iso"])
    .size()
    .reset_index(name="casos")
)


# ============================================================
# ORGANIZA OS DADOS EM UMA TABELA
# ============================================================

# Cada linha = uma semana
# Cada coluna = um ano

tabela = casos_semanais.pivot(
    index="semana_iso",
    columns="ano_iso",
    values="casos"
)

# Garante semanas 1 até 53
tabela = tabela.reindex(range(1, 54), fill_value=0)


# ============================================================
# GRÁFICO
# ============================================================

plt.figure(figsize=(16, 8))


for ano in tabela.columns:

    # Casos semanais
    casos = tabela[ano]

    # Média móvel de 8 semanas
    media_movel = casos.rolling(
        window=JANELA_MEDIA,
        min_periods=1
    ).mean()

    # Linha da média móvel
    plt.plot(
        tabela.index,
        media_movel,
        linewidth=2,
        label=str(ano)
    )


# ============================================================
# CONFIGURAÇÃO DO GRÁFICO
# ============================================================

plt.title(
    "Sazonalidade dos casos de HIV por semana\n"
    "Média móvel de 8 semanas"
)

plt.xlabel("Semana ISO")
plt.ylabel("Número médio de casos (8 semanas)")

plt.xticks(range(1, 54, 2))

plt.grid(
    True,
    alpha=0.3
)

plt.legend(
    title="Ano",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()
```
